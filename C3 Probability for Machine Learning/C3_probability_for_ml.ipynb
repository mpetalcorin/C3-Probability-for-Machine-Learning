# 🎲 Probability for Machine Learning: A Beginner's Guide

Probability is at the **heart of machine learning**. Whenever we talk about predicting outcomes, making decisions under uncertainty, or modeling randomness, we’re using **probability concepts**.

This notebook shows how probability works in **simple, real-world examples** and how it connects to machine learning tasks.

## What You'll Learn
- **Basic probability** (what are the chances of something happening?)
- **Conditional probability** (how do chances change when we know something else?)
- **Joint probability** (what’s the chance two things happen together?)
- **Bayes’ Theorem** (updating beliefs with new data)
- **Simulating probabilities** using Python

## Example 1: Probability of Flipping Heads 🪙
Let’s simulate flipping a fair coin 1000 times and estimate the probability of getting heads.

```python
import numpy as np

np.random.seed(42)

flips = np.random.choice(['Heads', 'Tails'], size=1000)
p_heads = np.mean(flips == 'Heads')
print(f"Estimated probability of Heads: {p_heads:.2f}")
```

## Example 2: Conditional Probability – Weather and Traffic 🌧️🚗
Imagine:
- Probability of rain: 30%
- Probability of traffic jam: 40%
- Probability of traffic jam **if it’s raining**: 70%

What’s the probability of a traffic jam **given that it rains?** Let’s check via simulation.

```python
num_days = 10000
rain = np.random.rand(num_days) < 0.3
traffic = np.zeros(num_days)

for i in range(num_days):
    if rain[i]:
        traffic[i] = np.random.rand() < 0.7
    else:
        traffic[i] = np.random.rand() < 0.4

p_traffic_given_rain = np.mean(traffic[rain])
print(f"Estimated P(Traffic | Rain): {p_traffic_given_rain:.2f}")
```

## Example 3: Joint Probability 🎯
What’s the probability that **both** it rains and there’s a traffic jam?

```python
p_rain_and_traffic = np.mean((rain) & (traffic == 1))
print(f"Estimated P(Rain and Traffic): {p_rain_and_traffic:.2f}")
```

## Example 4: Bayes’ Theorem 🧠
Let’s say:
- 1% of emails are spam
- 90% of spam emails contain the word “win”
- 5% of non-spam emails contain the word “win”

If we find an email with “win”, what’s the probability it’s spam? We’ll calculate this using **Bayes’ Theorem.**

```python
p_spam = 0.01
p_win_given_spam = 0.9
p_win_given_not_spam = 0.05

# Total probability of 'win'
p_win = p_spam * p_win_given_spam + (1 - p_spam) * p_win_given_not_spam

# Bayes’ Theorem
p_spam_given_win = (p_spam * p_win_given_spam) / p_win
print(f"P(Spam | 'win' found): {p_spam_given_win:.2f}")
```

## 🌟 Why This Matters for Machine Learning
- **Classification models** output probabilities (e.g., probability an image is a cat)
- **Naive Bayes classifiers** use Bayes’ Theorem
- **Uncertainty estimation** helps us know when the model is unsure

Probability helps us handle uncertainty, make predictions, and **train smarter models**.
