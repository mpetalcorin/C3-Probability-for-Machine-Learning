# NumPy in Science & Technology

NumPy is a **powerful Python library** that makes working with numbers and arrays super efficient. It’s like a **turbocharger for math in Python**.

What makes NumPy special? Instead of using slow Python loops, NumPy crunches numbers **really fast** behind the scenes using optimized C code.

I realized how empowering it is to simulate thousands of coin flips, dice rolls, 
or customer purchases in just a **few lines of code**—no more doing things by hand!

## What is NumPy?
NumPy stands for **Numerical Python**. It gives us:

- **Arrays** (like lists, but faster and more powerful)
- **Mathematical functions** (mean, sum, standard deviation, etc.)
- **Random simulations**

It’s widely used in **science, technology, finance, and engineering** because it makes large-scale number crunching super easy.

## Example 1: Simulating 10,000 Coin Flips
Let’s flip a coin 10,000 times and count how many heads and tails we get.

```python
import numpy as np

np.random.seed(42)

flips = np.random.choice(['Heads', 'Tails'], size=10000)
unique, counts = np.unique(flips, return_counts=True)
flip_results = dict(zip(unique, counts))
print(f"Results of 10,000 coin flips: {flip_results}")
```

## Example 2: Rolling 2 Dice 5,000 Times 
We’ll roll two 6-sided dice 5,000 times and see the distribution of their sums.

```python
dice1 = np.random.randint(1, 7, size=5000)
dice2 = np.random.randint(1, 7, size=5000)
sums = dice1 + dice2

import matplotlib.pyplot as plt

plt.hist(sums, bins=np.arange(2, 14)-0.5, edgecolor='black')
plt.title("Sum of Two Dice Rolled 5,000 Times")
plt.xlabel("Dice Sum")
plt.ylabel("Frequency")
plt.xticks(range(2, 13))
plt.show()
```

## Example 3: Simulating Customer Purchases 
Given 1,000 customers visit a store. Each spends between £5 and £100. We’ll simulate their purchases and calculate the total and average spent.

```python
purchases = np.random.uniform(5, 100, size=1000)
print(f"Total sales: £{purchases.sum():.2f}")
print(f"Average purchase: £{purchases.mean():.2f}")
```

## Example 4: Basic Array Math 
We’ll create an array of 10 numbers and perform basic math operations (addition, multiplication, etc.).

```python
array = np.arange(1, 11)
print("Original array:", array)
print("Array + 5:", array + 5)
print("Array squared:", array ** 2)
print("Square root:", np.sqrt(array))
```
