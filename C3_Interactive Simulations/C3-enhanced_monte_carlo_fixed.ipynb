# Monte Carlo Simulation – Visual 

In this notebook, we explore **Monte Carlo simulations** with visualizations and clear explanations. Monte Carlo simulations are all about using **random sampling** to estimate values or probabilities—like a virtual casino experiment.

**Applications:**
- Estimating π
- Risk analysis in finance
- Modeling randomness in science

## Estimating π Using Monte Carlo
We throw random points inside a square and check how many land inside a quarter circle.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style='whitegrid')
np.random.seed(42)
n_points = 10000
x = np.random.uniform(0, 1, n_points)
y = np.random.uniform(0, 1, n_points)
inside_circle = x**2 + y**2 <= 1
pi_estimate = (inside_circle.sum() / n_points) * 4
print(f'Estimated π: {pi_estimate:.5f}')
```

## Visual: Monte Carlo Points
We use **colors** to highlight points inside and outside the quarter circle.

```python
plt.figure(figsize=(6,6))
plt.scatter(x[inside_circle], y[inside_circle], color='lightblue', s=5, label='Inside Circle')
plt.scatter(x[~inside_circle], y[~inside_circle], color='salmon', s=5, label='Outside Circle')
circle = plt.Circle((0, 0), 1, color='black', fill=False)
plt.gca().add_patch(circle)
plt.title(f'Monte Carlo Simulation of π (N={n_points})')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.axis('square')
plt.show()
```

## Convergence Plot
This shows how the π estimate **stabilizes** as more points are used.

```python
estimates = []
for i in range(1, n_points+1):
    current_pi = (np.sum(x[:i]**2 + y[:i]**2 <= 1) / i) * 4
    estimates.append(current_pi)

plt.figure(figsize=(10,5))
plt.plot(estimates, label='π Estimate')
plt.axhline(np.pi, color='red', linestyle='--', label='True π')
plt.title('Convergence of π Estimate')
plt.xlabel('Number of Points')
plt.ylabel('Estimated π')
plt.legend()
plt.show()
```

## Dice Simulation: Probability of Rolling > 4
We simulate rolling two dice many times and estimate the probability of the total being greater than 4.

```python
n_simulations = 10000
die1 = np.random.randint(1, 7, n_simulations)
die2 = np.random.randint(1, 7, n_simulations)
totals = die1 + die2
prob_over_4 = np.mean(totals > 4)
print(f'Estimated Probability of Total > 4: {prob_over_4:.3f}')
```

## Visual: Dice Total Distribution
We show the **frequency distribution** of the dice totals.

```python
sns.histplot(totals, bins=np.arange(2,14)-0.5, kde=False, color='mediumseagreen')
plt.title('Distribution of Two Dice Totals (10,000 Simulations)')
plt.xlabel('Total')
plt.ylabel('Frequency')
plt.show()
```
