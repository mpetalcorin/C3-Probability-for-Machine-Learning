# Monte Carlo Simulation with Interactive Widgets & Real-World Data

In this notebook, we:
- Add **interactive widgets** for real-time simulations
- Incorporate a **real-world drug discovery dataset**
- Provide **dynamic visualizations**

## Interactive π Estimation
Use the slider below to set the number of simulation points.

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

def monte_carlo_pi(n_points=1000):
    x = np.random.uniform(0, 1, n_points)
    y = np.random.uniform(0, 1, n_points)
    inside_circle = x**2 + y**2 <= 1
    pi_estimate = (inside_circle.sum() / n_points) * 4

    plt.figure(figsize=(5,5))
    plt.scatter(x[inside_circle], y[inside_circle], color='lightblue', s=3, label='Inside')
    plt.scatter(x[~inside_circle], y[~inside_circle], color='salmon', s=3, label='Outside')
    plt.title(f'π Estimate: {pi_estimate:.5f} (N={n_points})')
    plt.legend()
    plt.show()

widgets.interact(monte_carlo_pi, n_points=widgets.IntSlider(1000, 100, 20000, step=500))
```

## Interactive Dice Simulation
Choose the number of dice and number of rolls.

```python
def dice_simulation(n_dice=2, n_rolls=1000):
    totals = np.sum(np.random.randint(1, 7, size=(n_rolls, n_dice)), axis=1)
    plt.figure(figsize=(8,4))
    plt.hist(totals, bins=range(n_dice, n_dice*6+2), color='mediumseagreen', edgecolor='black')
    plt.title(f'Distribution of Totals for {n_dice} Dice ({n_rolls} Rolls)')
    plt.xlabel('Total')
    plt.ylabel('Frequency')
    plt.show()

widgets.interact(
    dice_simulation,
    n_dice=widgets.IntSlider(2, 1, 5, step=1),
    n_rolls=widgets.IntSlider(1000, 500, 10000, step=500)
)
```

## Real-World Dataset: Drug Discovery Simulation
We simulate the probability that a compound from a dataset exceeds an **activity threshold**.

```python
import pandas as pd
import seaborn as sns
# Create a synthetic drug discovery dataset
np.random.seed(42)
n_compounds = 1000
data = pd.DataFrame({
    'Compound_ID': [f'CMPD_{i}' for i in range(n_compounds)],
    'Activity': np.random.normal(loc=5, scale=2, size=n_compounds)
})

def activity_simulation(threshold=6.0):
    hits = data['Activity'] >= threshold
    prob = hits.mean()
    sns.histplot(data['Activity'], bins=30, color='cornflowerblue')
    plt.axvline(threshold, color='red', linestyle='--', label=f'Threshold = {threshold}')
    plt.title(f'Probability of Activity ≥ {threshold}: {prob:.2%}')
    plt.xlabel('Activity')
    plt.ylabel('Count')
    plt.legend()
    plt.show()

widgets.interact(activity_simulation, threshold=widgets.FloatSlider(value=6.0, min=2.0, max=10.0, step=0.1))
```
