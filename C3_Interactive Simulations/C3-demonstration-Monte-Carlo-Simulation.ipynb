```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# from scipy.stats import pearsonr

def simulate_walk(n, p=0.6):
    sim = np.zeros(n)
    for _ in range(n):
        sim[_] = 1 if np.random.rand() < p else -1
    return sim
        
walk_data = simulate_walk(100)
walk_data[:10]
```

```python
sum(walk_data) / 100
```

```python
def simulate_random_walk(n_paths, path_length, trend=0.5):
    sim = np.zeros((n_paths, path_length))
    for _ in range(n_paths):
        path = simulate_walk(path_length, trend)
        sim[_, :]= np.cumsum(path)
    return sim

random_walk = simulate_random_walk(100, 100)

fig, ax = plt.subplots()
for p in random_walk:
    ax.plot(p, alpha=0.4)
```

```python
# area of a disk = pi * radius^2
def estimate_pi(n=10000):
    count = 0
    for _ in range(n):
        x, y = np.random.rand(2)
        if x**2 + y**2 < 1:
            count += 1
    return (count / n) / (0.5)**2

estimate_pi()
```
